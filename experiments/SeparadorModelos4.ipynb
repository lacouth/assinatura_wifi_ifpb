```python
import pandas as pd
```

```python
# Arquivo de entrada
arquivo_entrada = "../data/verification/PredicaoModelos080526.csv"

# Lê o CSV
df = pd.read_csv(arquivo_entrada)

# Remove a primeira coluna caso seja apenas índice
if df.columns[0] not in ["data", "hora", "num", "ambiente"]:
    df = df.iloc[:, 1:]

# Lista dos modelos
modelos = ["XGB05", "XGB10", "RF05", "RF10"]

# Para cada modelo, cria um CSV separado
for modelo in modelos:

    # Seleciona apenas as colunas desejadas
    df_modelo = df[["data", "hora", "num", "ambiente", modelo]].copy()

    # Renomeia a última coluna para minúsculo
    df_modelo.columns = [
        "data",
        "hora",
        "num",
        "ambiente",
        modelo.lower()
    ]

    # Nome do arquivo de saída
    arquivo_saida = f"{modelo.lower()}.csv"

    # Salva o CSV
    df_modelo.to_csv(arquivo_saida, index=False)

    print(f"Arquivo gerado: {arquivo_saida}")
```

## **Importação das Bibliotecas**

```python
import pandas as pd

# Arquivos
arquivo_entrada = "../data/verification/experiments/VerificacaoModelos.csv"
arquivo_saida = "../data/verification/ModelosTaguiados.csv"

# Modelos na ordem desejada
modelos = ["XGB05", "XGB10", "RF05"]

# Leitura do CSV
df = pd.read_csv(arquivo_entrada)

# Criar lista de modelos repetindo de 3 em 3
df["modelo"] = [modelos[i % 3] for i in range(len(df))]

# Definir ordem personalizada
df["modelo"] = pd.Categorical(df["modelo"], categories=modelos, ordered=True)

# Ordenar pelo modelo
df = df.sort_values(by="modelo").reset_index(drop=True)

# Salvar novo arquivo
df.to_csv(arquivo_saida, index=False)

print("Arquivo gerado com sucesso!")
```

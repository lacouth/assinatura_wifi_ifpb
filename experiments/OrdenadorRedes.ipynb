```python
import pandas as pd
```

```python
# ===== CONFIGURAÇÃO =====
arquivo_entrada = "../data/redes2.csv"          # CSV original
arquivo_saida = "../data/redes2_atualizado.csv" # CSV de saída

incremento = 150
```

```python
# ===== LEITURA =====
df = pd.read_csv(arquivo_entrada)

# ===== GARANTE QUE A COLUNA É NUMÉRICA =====
df["num"] = pd.to_numeric(df["num"], errors="coerce")

# ===== SOMA 150 NA COLUNA =====
df["num"] = df["num"] + incremento

# ===== SALVAR =====
df.to_csv(arquivo_saida, index=False)

print(f"Arquivo salvo como: {arquivo_saida}")
```

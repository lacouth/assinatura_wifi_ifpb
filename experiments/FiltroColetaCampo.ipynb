## **Importação das bibliotecas**

```python
import pandas as pd
```

## **Leitura do Arquivo CSV**

```python
# Arquivos
arquivo_entrada = "../data/verification/experiments/VerificacaoMicroComColetaXGBsRF.csv"
arquivo_saida = "../data/verification/experiments/VerificacaoMicroComColetaXGBsRFSaida.csv"

# Ler CSV
df = pd.read_csv(arquivo_entrada)
```

## **Filtro**

```python
# Marcar a última linha de cada grupo
df["remover"] = df.groupby("num").cumcount(ascending=False) == 0

# Filtrar (remover essas linhas)
df_filtrado = df[df["remover"] == False].drop(columns="remover")

# Resetar índice (opcional)
df_filtrado = df_filtrado.reset_index(drop=True)

# Salvar novo CSV
df_filtrado.to_csv(arquivo_saida, index=False)

print("Arquivo salvo com sucesso!")
```

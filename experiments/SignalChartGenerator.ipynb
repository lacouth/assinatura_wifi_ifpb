## **Importação das Bibliotecas**

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import os
```

## **Configuração e Leitura do Arquivo CSV**

```python
arquivo_csv = "../data/signs/AgrupamentoDeRedes.csv"
pasta_saida = "../images/SignalChart"

df = pd.read_csv(arquivo_csv)
```

## **Conversão e Preparação**

```python
# Converter tipos
df["Minimo"] = df["Minimo"].astype(float)
df["Maximo"] = df["Maximo"].astype(float)
df["Quantidade"] = df["Quantidade"].astype(int)

# ===== PREPARAÇÃO =====
df["Range"] = df["Maximo"] - df["Minimo"]
```

## **Função de Cor**

```python
def definir_cor(qtd):
    if qtd >= 8:
        return "green"
    elif 6 <= qtd < 8:
        return "lightblue"
    elif qtd == 5:
        return "yellow"
    else:
        return "red"
```

## **Gerador Gráfico - Por Ambiente**

```python
for local, df_local in df.groupby("Local"):

    # ===== ORDENAR ROTEADORES =====
    df_local = df_local.sort_values(
        by="NumRoteador",
        key=lambda x: x.str.extract(r'(\d+)').astype(int)[0]
    )

    cores = df_local["Quantidade"].apply(definir_cor)

    plt.figure(figsize=(10, 6))

    plt.barh(
        df_local["NumRoteador"],
        df_local["Range"],
        left=df_local["Minimo"],
        color=cores
    )

    plt.xlabel("Sinal (dBm)")
    plt.ylabel("Roteador")
    plt.title(f"{local} - Intervalo de Sinal (Min x Max)")

    plt.gca().invert_yaxis()

    # ===== FIXAR RANGE DO EIXO X =====
    plt.xlim(-100, -45)

    # ===== LEGENDA =====
    legenda = [
        Patch(color='green', label='≥ 8 leituras'),
        Patch(color='lightblue', label='6–7 leituras'),
        Patch(color='yellow', label='5 leituras'),
        Patch(color='red', label='< 5 leituras')
    ]

    plt.legend(handles=legenda, title="Quantidade de leituras")
    plt.tight_layout()

    # Limpar nome do arquivo (remover caracteres problemáticos)
    nome_arquivo = local.replace(" ", "_").replace(".", "").replace("ç", "c")
    caminho_saida = os.path.join(pasta_saida, f"{nome_arquivo}.png")

    # ===== SALVAR =====
    plt.savefig(caminho_saida, dpi=300)
    plt.close()

    print(f"Gráfico salvo em: {caminho_saida}")
```

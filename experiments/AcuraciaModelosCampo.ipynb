````python
import pandas as pd
from sklearn.metrics import accuracy_score
````

````python
# Caminho do CSV

# ===== Testes em Campo =====
name = "RF05EE080526"
#name = "RF10EE080526"
#name = "XGB05EE080526"
#name = "XGB10EE080526"

arquivo = "../data/verification/PredicaoCampo/" + name + ".csv"  # <-- altera aqui

# ===== Testes em Campo Virtual =====
#name = "RF05EE080526VC"
#name = "RF10EE080526VC"
#name = "XGB05EE080526VC"
#name = "XGB10EE080526VC"

#arquivo = "../data/verification/PredicaoCampoVirtual/" + name + ".csv"  # <-- altera aqui
````

````python
# Lê o arquivo
df = pd.read_csv(arquivo)

# Remove a primeira coluna caso seja apenas índice
if df.columns[0] not in ["data", "hora", "num", "ambiente", "predição"]:
    df = df.iloc[:, 1:]

# Valores reais
y_true = df["ambiente"]

# Valores preditos
y_pred = df["predição"]

# Calcula a acurácia
acuracia = accuracy_score(y_true, y_pred)

# Exibe o resultado
print(f"Acurácia: {acuracia:.4f}")
print(f"Acurácia (%): {acuracia * 100:.2f}%")
````

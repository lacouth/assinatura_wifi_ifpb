## **Importação das bibliotecas**

```python
import pandas as pd
```

## **Configuração**

```python
arquivo_entrada = "../data/source/PS1024EE05.csv"   # Caminho para o CSV original
arquivo_saida = "../data/source/FS1024EE05.csv"  # Caminho para o CSV filtrado

# ===== FILTROS =====
ssid_manter = ["IFPB", "Visitante"]   # mantém se contiver QUALQUER um desses
ssid_remover = ["Sala", "TCC", "MM"]  # remove se contiver QUALQUER um desses
```

## **Leitura e Filtro**

```python
# ===== LEITURA =====
df = pd.read_csv(arquivo_entrada)

# ===== FILTRO DE MANUTENÇÃO =====
filtro_manter = df["ssid"].str.contains("|".join(ssid_manter), case=False, na=False)

# ===== FILTRO DE REMOÇÃO =====
filtro_remover = df["ssid"].str.contains("|".join(ssid_remover), case=False, na=False)

# ===== APLICAÇÃO FINAL =====
df_filtrado = df[filtro_manter & ~filtro_remover]

# ===== SALVAR =====
df_filtrado.to_csv(arquivo_saida, index=False)

print(f"Arquivo filtrado salvo como: {arquivo_saida}")
print(f"Total de linhas mantidas: {len(df_filtrado)}")
```

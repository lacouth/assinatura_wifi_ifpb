```python
import pandas as pd

# Arquivo de entrada
arquivo_entrada = "../data/source/PS080526EE10.csv"

# Arquivo de saída
arquivo_saida = "../data/verification/PredicaoModelos080526.csv"

# Lê o CSV original
df = pd.read_csv(arquivo_entrada)

# Seleciona apenas a última linha de cada número "num"
df_final = df.groupby("num", as_index=False).last()

# Renomeia as colunas
df_final.columns = [
    "data",
    "hora",
    "num",
    "ambiente",
    "modelo1",
    "modelo2",
    "modelo3",
    "modelo4"
]

# Salva o novo CSV
df_final.to_csv(arquivo_saida, index=False)

print(f"Arquivo salvo com sucesso: {arquivo_saida}")
```

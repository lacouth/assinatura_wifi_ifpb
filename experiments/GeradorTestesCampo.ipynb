```python
import pandas as pd

# Arquivos
arquivo_entrada = "../data/source/FSMicroColetaModelos.csv"
arquivo_saida = "../data/dataset/FSMicroColetaModelos.csv"

# Ler CSV
df = pd.read_csv(arquivo_entrada)

# Garantir MAC em minúsculo
df["bssid"] = df["bssid"].astype(str).str.lower()

# Criar tabela pivô com num e local
tabela = df.pivot_table(
    index=["num", "local"],   # 👈 agora tem dois índices
    columns="bssid",
    values="rssi",
    aggfunc="first"
)

# Ordenar colunas (MACs)
tabela = tabela.reindex(sorted(tabela.columns), axis=1)

# Resetar índice (vira colunas novamente)
tabela = tabela.reset_index()

# (Opcional) preencher valores ausentes
tabela = tabela.fillna(-100)

# Salvar CSV
tabela.to_csv(arquivo_saida, index=False)

print("Arquivo salvo com sucesso!")

print(tabela)
```

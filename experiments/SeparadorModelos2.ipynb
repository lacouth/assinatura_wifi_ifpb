## **Importação das bibliotecas**

```python
import pandas as pd
import os
```

## **Leitura do Arquivo CSV**

```python
# Arquivos
arquivo_entrada = "../data/verification/experiments/VerificacaoModelos.csv"
pasta_saida = "../data/verification/"

# Modelos na ordem desejada
modelos = ["XGB05", "XGB10", "RF05"]

# Leitura do CSV
df = pd.read_csv(arquivo_entrada)
```

## **Separador de Modelos**

```python
# Criar coluna de modelo (cíclico)
df["modelo"] = [modelos[i % 3] for i in range(len(df))]

# ===== SALVAR EM 3 ARQUIVOS =====
for modelo in modelos:

    df_modelo = df[df["modelo"] == modelo]

    caminho_saida = os.path.join(pasta_saida, f"{modelo}.csv")

    df_modelo.to_csv(caminho_saida, index=False)

    print(f"Arquivo salvo: {caminho_saida}")
```
